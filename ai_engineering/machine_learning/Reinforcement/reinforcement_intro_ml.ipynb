# **Reinforcement Algorithms**

- Q-learning
    - Model-free, off-policy, value-based RL.
    - Learns Q(s, a): “how good is action a in state s?”
    - Your notebook is using this directly.

- SARSA
    - Very similar to Q-learning, but on-policy.
    - Updates using the action actually taken, not the greedy one.

- Deep Q-Network (DQN)
    - Q-learning with a neural network as the Q-function approximator.
    - Used for larger state spaces like Atari games.

- Policy Gradient / REINFORCE
    - Directly learns a policy π(a|s) instead of a Q-table.
    - Good when actions are continuous or state/action spaces are huge.

- Actor-Critic
    - Combines value estimation (critic) with policy learning (actor).
    - Common modern family: A2C, A3C, PPO, SAC, TD3.

- Dynamic Programming
    - Value iteration and policy iteration.
    - Requires a known model of the environment.

- Model-based RL
    - Learns or uses a model of the environment to plan.
    - Examples: Dyna, MPC-style planning.

In plain terms:
- Q-learning is the classic tabular RL algorithm for small, discrete environments.
- SARSA is similar but more conservative because it follows the current policy.
- DQN, PPO, SAC are modern extensions for larger, more complex tasks.

```python
import numpy as np

# Simple 4x4 grid: start at (0, 0), goal at (3, 3)
n_states = 16
n_actions = 4  # up, right, down, left
goal = 15
q_table = np.zeros((n_states, n_actions))

moves = [(-1, 0), (0, 1), (1, 0), (0, -1)]

def step(state, action):
    row, col = divmod(state, 4)
    dr, dc = moves[action]
    row = np.clip(row + dr, 0, 3)
    col = np.clip(col + dc, 0, 3)
    next_state = row * 4 + col
    reward = 10 if next_state == goal else -1
    return next_state, reward

# Train with Q-learning
learning_rate = 0.1
discount = 0.95
epsilon = 0.2

for episode in range(1000):
    state = 0

    for _ in range(100):
        if np.random.random() < epsilon:
            action = np.random.randint(n_actions)
        else:
            action = np.argmax(q_table[state])

        next_state, reward = step(state, action)
        q_table[state, action] += learning_rate * (
            reward + discount * np.max(q_table[next_state]) - q_table[state, action]
        )
        state = next_state

        if state == goal:
            break

# Display the learned best action for each cell
action_symbols = ["↑", "→", "↓", "←"]
for state in range(n_states):
    print("G" if state == goal else action_symbols[np.argmax(q_table[state])], end=" ")
    if state % 4 == 3:
        print()
```
