# Supervised Machine Learning
 - Supervised machine learning trains a model using examples that include both input data and the correct answers (labels). The model learns to predict answers for new data. Common tasks include classification and regression.

## Major supervised learning algorithms
### 1. Regression:
 - Linear and polynomial regression
 - Decision tree 
 - Random forest regression
 - Support vector regression (SVR)
 - Gradient boosting regression

### 2.  Classification:
 - Logistic regression
 - Decision trees 
 - Random forests
 - Support vector machines (SVM)
 - k-nearest neighbors (k-NN)
 - Naive Bayes
 - Gradient boosting (e.g., XGBoost, LightGBM)

```python
import numpy as np
from sklearn.linear_model import LinearRegression

import matplotlib.pyplot as plt

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 6])

# Train the model
model = LinearRegression()
model.fit(X, y)

# Predict and display the fitted line
y_pred = model.predict(X)
print(f"Slope: {model.coef_[0]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")

plt.scatter(X, y, label="Data")
plt.plot(X, y_pred, color="red", label="Regression line")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor(random_state=42)
tree_model.fit(X, y)

tree_pred = tree_model.predict(X)

plt.scatter(X, y, label="Data")
plt.plot(X, tree_pred, color="green", marker="o", label="Decision tree")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

forest_model = RandomForestRegressor(n_estimators=100, random_state=42)
forest_model.fit(X, y)

forest_pred = forest_model.predict(X)

plt.scatter(X, y, label="Data")
plt.plot(X, forest_pred, color="purple", marker="o", label="Random forest")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
from sklearn.svm import SVR

svr_model = SVR(kernel="linear", C=1000, epsilon=0.1)
svr_model.fit(X, y)

svr_pred = svr_model.predict(X)

plt.scatter(X, y, label="Data")
plt.plot(X, svr_pred, color="orange", marker="o", label="Support vector regression")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(n_estimators=100, random_state=42)
gb_model.fit(X, y)

gb_pred = gb_model.predict(X)

plt.scatter(X, y, label="Data")
plt.plot(X, gb_pred, color="brown", marker="o", label="Gradient boosting")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X_cls, y_cls = make_classification(
    n_samples=200,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X_cls, y_cls, test_size=0.25, random_state=42, stratify=y_cls
)

logistic_model = LogisticRegression()
logistic_model.fit(X_train, y_train)

y_cls_pred = logistic_model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=y_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Logistic Regression Predictions")
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_classifier = DecisionTreeClassifier(random_state=42)
tree_classifier.fit(X_train, y_train)

tree_cls_pred = tree_classifier.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, tree_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=tree_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Tree Classifier Predictions")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

forest_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
forest_classifier.fit(X_train, y_train)

forest_cls_pred = forest_classifier.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, forest_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=forest_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Random Forest Classifier Predictions")
plt.show()
```

```python
from sklearn.svm import SVC

svc_classifier = SVC(kernel="rbf", random_state=42)
svc_classifier.fit(X_train, y_train)

svc_cls_pred = svc_classifier.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, svc_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=svc_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Support Vector Classifier Predictions")
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_classifier = KNeighborsClassifier(n_neighbors=5)
knn_classifier.fit(X_train, y_train)

knn_cls_pred = knn_classifier.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, knn_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=knn_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("k-Nearest Neighbors Predictions")
plt.show()
```

```python
from sklearn.naive_bayes import GaussianNB

nb_classifier = GaussianNB()
nb_classifier.fit(X_train, y_train)

nb_cls_pred = nb_classifier.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, nb_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=nb_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Naive Bayes Predictions")
plt.show()
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_classifier = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb_classifier.fit(X_train, y_train)

gb_cls_pred = gb_classifier.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, gb_cls_pred):.2f}")

plt.scatter(X_test[:, 0], X_test[:, 1], c=gb_cls_pred, cmap="coolwarm")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Gradient Boosting Classifier Predictions")
plt.show()
```
