# Generative AI

Generative AI creates new content. Large language models (LLMs) are generative AI models focused on language.

## Major tasks in generative AI

- **Text generation:** writing, summarization, translation, and question answering.
- **Code generation:** writing, explaining, and completing code.
- **Image generation and editing:** creating or modifying images from prompts.
- **Audio and speech generation:** producing speech, music, or sound; transcription and voice synthesis.
- **Video generation:** creating or editing video.
- **Multimodal generation:** understanding and generating combinations of text, images, audio, and video.

## Important model families

- **Transformers:** the architecture behind many language models, including GPT, Claude, Gemini, and Llama.
- **Diffusion models:** widely used for image generation, including Stable Diffusion and Imagen.
- **GANs (Generative Adversarial Networks):** an earlier approach to generating realistic images and other data.
- **VAEs (Variational Autoencoders):** models that learn compact representations and generate new samples.
- **Multimodal models:** models such as GPT-4o and Gemini that work across multiple types of data.

There is no single “most important” model; the best choice depends on the task, quality requirements, cost, and available resources.

## Prompt engineering

Prompt engineering is the practice of designing and refining instructions to guide a generative AI model toward useful, relevant, and reliable responses.

### Helpful techniques

- **Be specific:** State the task, audience, tone, and desired level of detail.
- **Provide context:** Include relevant background, data, or examples.
- **Set constraints:** Specify format, length, scope, or what to avoid.
- **Use examples:** Show the model what a good response should look like.
- **Break down complex tasks:** Ask for clear steps or handle subtasks separately.
- **Iterate:** Review the output and refine the prompt when needed.

### Example

**Vague prompt:**  
“Explain climate change.”

**More specific prompt:**  
“Explain the main causes of climate change in three short paragraphs for a middle-school audience. Use plain language and include one example of how it affects daily life.”

## AI agents

An AI agent is a system that uses an AI model to pursue a goal by observing information, making decisions, and taking actions. Unlike a model that only responds to a prompt, an agent may use tools, retain relevant context, and repeat steps until it completes a task or reaches a limit.

### Common components

- **Model:** interprets the task and helps decide what to do.
- **Tools:** let the agent interact with services, files, or other systems.
- **Memory or state:** keeps track of relevant context and progress.
- **Control loop:** selects actions, checks results, and decides whether to continue.

### Example workflow

1. Receive a goal, such as finding and summarizing recent research.
2. Search using an available tool.
3. Review the results and gather more information if needed.
4. Produce a summary and report what sources were used.

Agents can automate multi-step tasks, but their actions should be bounded and reviewed especially when they affect external systems or important data.
