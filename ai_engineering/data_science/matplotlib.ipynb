```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [2, 4, 3])
plt.title("A simple line plot")
plt.show()
```

```python
from matplotlib_venn import venn2

set_a = {"A", "B", "C"}
set_b = {"B", "C", "D"}

venn2([set_a, set_b], set_labels=("Set A", "Set B"))
plt.title("Venn Diagram")
plt.show()
```

```python
plt.bar(["A", "B", "C"], [3, 5, 2])
plt.title("Example bar chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

```python
values = [2, 3, 3, 4, 5, 5, 5, 6, 7, 8]
plt.hist(values, bins=5, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Example data
x = [1, 2, 3, 4, 5, 6]
y = [3, 5, 2, 7, 6, 8]

# Create scatter plot
plt.figure(figsize=(6, 4))
plt.scatter(x, y, color="dodgerblue", s=80, alpha=0.8)

plt.title("Scatter Plot Example")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
negative_y = [8, 6, 7, 4, 3, 1]
no_trend_y = [4, 8, 2, 7, 1, 5]

fig_trends, axes_trends = plt.subplots(1, 3, figsize=(12, 4))

axes_trends[0].scatter(x, y, color="seagreen")
axes_trends[0].set_title("Positive trend")

axes_trends[1].scatter(x, negative_y, color="tomato")
axes_trends[1].set_title("Negative trend")

axes_trends[2].scatter(x, no_trend_y, color="slateblue")
axes_trends[2].set_title("No clear trend")

for axis in axes_trends:
    axis.set_xlabel("x")
    axis.set_ylabel("y")
    axis.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
