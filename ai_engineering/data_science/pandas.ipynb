```python
import pandas as pd

# Create a DataFrame
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["Paris", "Rome", "Berlin"]
})

print(df)

# Filter rows
print(df[df["Age"] > 28])

# Compute summary
print(df["Age"].mean())
```

```python
df.head()
df.info()
df.describe(include="all")
df.loc[df["Age"] > 28, ["Name", "Age"]]
df.sort_values("Age")
df.groupby("City")["Age"].mean()
df.isna().sum()
```
