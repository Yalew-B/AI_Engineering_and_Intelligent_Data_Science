```python
import numpy as np

# Create a NumPy array
arr = np.array([1, 2, 3, 4])

# Perform vectorized operations
print(arr * 2)       # [2 4 6 8]
print(arr.mean())   # 2.5
print(arr.sum())    # 10
```

```python
# Major NumPy tasks you should know
print("Original array:", arr)
print("Shape:", arr.shape)
print("Data type:", arr.dtype)

# 1) Indexing and slicing
print("First element:", arr[0])
print("Slice:", arr[1:3])

# 2) Vectorized operations
print("Multiply by 2:", arr * 2)
print("Add 10:", arr + 10)

# 3) Aggregations
print("Mean:", arr.mean())
print("Sum:", arr.sum())
print("Min:", arr.min())
print("Max:", arr.max())

# 4) Boolean filtering
print("Values greater than 2:", arr[arr > 2])

# 5) Reshaping
print("Reshaped 2x2:", arr.reshape(2, 2))

# 6) Sorting and argmax
print("Sorted:", np.sort(arr))
print("Index of max:", arr.argmax())

# 7) Broadcasting
print("Broadcasting example:", arr + np.array([10, 20, 30, 40]))
```
