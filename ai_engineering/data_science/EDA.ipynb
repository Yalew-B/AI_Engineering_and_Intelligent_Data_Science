EDA stands for Exploratory Data Analysis.

It is the process of examining a dataset to understand its structure, quality, and patterns before building models or making decisions. In data science, EDA helps answer questions like:

- What does the data look like?
- Are there missing or incorrect values?
- Are there outliers or unusual patterns?
- How are variables related to each other?
- What trends or distributions should I be aware of?

Typical EDA steps:
- Summary statistics: mean, median, std, min, max
- Missing value checks
- Data type checks
- Visualizations: histograms, box plots, scatter plots, bar charts, correlation heatmaps
- Detecting anomalies and inconsistencies

Why EDA matters:
- It helps clean and prepare data
- It reveals important patterns and relationships
- It reduces the risk of building a poor or misleading model
- It guides feature selection and modeling decisions

In short, EDA is the “first look” at the data to understand what is in it and what needs attention before analysis or machine learning.

# Major EDA Tasks You Should Know

1. **Understand the dataset**
    - Check its number of rows and columns.
    - Inspect column names, data types, and sample records.

2. **Assess data quality**
    - Find missing values and duplicates.
    - Check for invalid, inconsistent, or unexpected values.

3. **Summarize the data**
    - Calculate descriptive statistics.
    - Examine counts for categorical variables.

4. **Explore individual variables**
    - Plot distributions with histograms, bar charts, or box plots.
    - Identify skewness and potential outliers.

5. **Explore relationships**
    - Compare variables with scatter plots, grouped summaries, or cross-tabulations.
    - Check correlations between numerical variables.

6. **Investigate patterns and anomalies**
    - Look for trends, clusters, unusual observations, and class imbalance.

7. **Prepare for further analysis**
    - Decide how to handle missing values and outliers.
    - Identify useful features and any required data transformations.

```python
import pandas as pd

import matplotlib.pyplot as plt

# Sample dataset
df = pd.DataFrame({
    "age": [23, 35, 29, 42, 31, 35, None, 28],
    "income": [42000, 68000, 51000, 90000, 59000, 68000, 47000, 50000],
    "department": ["Sales", "IT", "HR", "IT", "Sales", "IT", "HR", "Sales"],
    "satisfaction": [7, 8, 6, 9, 7, 8, 5, 6]
})

# 1. Inspect structure and sample rows
print("Shape:", df.shape)
display(df.head())
df.info()

# 2. Check data quality
print("Missing values per column:")
display(df.isna().sum())

print("Duplicate rows:", df.duplicated().sum())

# 3. Summarize numerical and categorical columns
display(df.describe())
display(df.describe(include="object"))

# 4. Count categorical values
display(df["department"].value_counts())

# 5. Plot numerical distributions
df[["age", "income", "satisfaction"]].hist(figsize=(10, 6), bins=5)
plt.tight_layout()
plt.show()

# 6. Compare numerical variables
df.plot.scatter(x="age", y="income", title="Age vs. Income")
plt.show()

# 7. Explore correlations
correlation = df.select_dtypes(include="number").corr()
display(correlation)

plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
